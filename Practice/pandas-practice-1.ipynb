```python

import pandas as pd

data = {
    "Name": ["Faith", "Wahu", "Mary", "John", "Kevin", "Amina"],
    "Age": [20, 21, 19, 24, 22, 20],
    "Course": ["CS", "Architecture", "Art", "CS", "IT", "CS"],
    "Marks": [78, 85, 67, 90, 72, 88]
}

students = pd.DataFrame(data)

print(students)
#Display the first five rows of students. 
students.head()
#Find the number of rows and columns using one line of code. Then write a comment explaining what each number means.
students.shape#(6,4) meaning 6 rows and 4 columns
#Display all the column names in students.
print(students.columns)
#Display only the Name column. Then display only the Marks column.
print(students["Name"])
print(students["Marks"])
#Use iloc to display the third student's entire row. 
print(students.iloc[2])
#Then select that student's Marks value only.
print("Third's students marks",students.iloc[2,3])
#Display only students whose age is greater than 20.
print(students[students["Age"]>20])
#Display only students taking CS.
print(students[students["Course"]=="CS"])
#Display students who take CS and have marks greater than 80.
print(students[(students["Course"]=="CS")&(students["Marks"]>80)])
#Display all students sorted from the highest marks to the lowest marks.
print(students.sort_values(by=["Marks"],ascending=False))
#Add a column called Result:
#- "Pass" if marks are 50 or higher.
#- "Fail" otherwise.
#use a simple list of six values that matches the students' marks.
students["Result"]=["Pass","Pass","Pass","Pass","Pass","Pass"]
print(students)
#Find the number of missing values in each column.
print(students.isnull().sum())
#Then check whether the Age column contains any missing values.
print("The number of missing values in age column:",students["Age"].isnull().sum())
#Save your DataFrame as students_practice.csv without including the index as an extra column.
students.to_csv("students_practice.csv",index=False)
#Then read the CSV back into a new variable called students_loaded and display it.
students_loaded=pd.read_csv("students_practice.csv")
print(students_loaded)
#Create a new column called Grade using the Marks column:
#- Marks of 80 or higher → "A"
#- Marks from 70 to 79 → "B"
#- Marks below 70 → "C"
def assign_grade(mark):#def is used to define a function
    if mark >= 80:
        return "A"
    elif mark >= 70:
        return "B"
    else:
        return "C"
students_loaded["Grade"] = students_loaded["Marks"].apply(assign_grade)#apply means the function should be applied to each value(Marks)

print(students_loaded)





```
