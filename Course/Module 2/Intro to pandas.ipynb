```python
#Task 1 — Create a Series
#Create a Pandas Series containing the following ages:20, 21, 19, 22, 20, 23
import pandas as pd
ages=pd.Series([20, 21, 19, 22, 20, 23])

#Then:
#Print the Series.
print(ages)
#Print its index.
print(ages.index)
#Print its values.
print(ages.values)
#Print the first value.
print(ages[0])

#Find the average age.
ages.mean()
print("the average for the ages is:",ages.mean())
```

```python
#Task 2 — Create your first DataFrame
data={"Name":["Faith","Wahu","Ndung'u"],"Age":[20,21,22],"Course":["CS","Archi","Art"]}
students=pd.DataFrame(data)
print(students)

#Task 3 — Selecting data from a DataFrame
#Using your existing students DataFrame, try these four tasks:
#1. Print only the Name column.
print(students["Name"])
#2. Print only the Age column.
print(students["Age"])
#3. Print the first student's entire row.
print(students.iloc[0])
#4. Print the age of the second student.
print("Age of the second student:",students.iloc[1,1])
#print only the Course of the third student
print("Course of the third student:", students.iloc[2, 2])

#Task 4 — Filtering a DataFrame
#Display only students whose age is greater than 20.
print(students[students["Age"]>20])
#Display only students whose course is "CS".
print([students[students["Course"]=="CS"]])

#Task 5 — Combining conditions
#students who are older than 20 AND taking Art.
print(students[(students["Age"]>20)&(students["Course"]=="Art")])

#Task 6 — Sorting your DataFrame
#Sort students by age from youngest to oldest.
print(students.sort_values(by="Age"))
#Sort students by age from oldest to youngest.
print(students.sort_values(by="Age",ascending=False))

#Task 7 — Getting information about your data
#Display the number of rows and columns.
print("Number of rows and columns:",students.shape)
#Display the column names.
print("column names:",students.columns)
#Display a summary of the DataFrame, including its column names, non-null counts, and data types.
students.info()#.info() prints its own summary and returns None. You don't need to wrap it in print().

#Task 8 — Missing values and basic statistics
#Display the number of missing values in each column.
print(students.isnull())#Checks every cell and returns True if the value is missing
print(students.isnull().sum())#prints the count of how many missing values there are in each column
#Calculate the average age.
print(" the average age:",students["Age"].mean())

#Task 9 — Let's practise changing data
#Add a new column called Grade with the values "A", "B", and "A" for the three students, respectively.
students["Grade"]=["A","B","A"]
#Change Faith's age from 20 to 23.
students.iloc[0,1]=23
print(students)

#Task 10 — Removing a column
students.drop(columns=["Grade"],inplace=True)#inplace=True makes changes to the current dataframe
print(students)

#Task 11 — Saving your DataFrame
students.to_csv("students.csv",index=False)
students_loaded = pd.read_csv("students.csv")
print(students_loaded)
```
